```python
import os
import requests

# Configuración actualizada para la Versión 3
DATASET_ID = "xw5xc7xrmp"
VERSION = "3"
DOWNLOAD_DIR = "../datasets/HVDROPDB_Original"
```

```python
os.makedirs(DOWNLOAD_DIR, exist_ok=True)
```

```python
def descargar_mendeley_v3():
    print(f"Conectando al dataset {DATASET_ID} (Versión {VERSION}) en Mendeley Data...")
    
    # Endpoint de la API pública de Mendeley para la versión específica
    api_url = f"https://data.mendeley.com/public-api/datasets/{DATASET_ID}/files?version={VERSION}"
    
    try:
        response = requests.get(api_url)
        response.raise_for_status()
        archivos = response.json()
        print(f"Se encontraron {len(archivos)} archivos en la Versión {VERSION}.")

        for archivo in archivos:
            nombre_archivo = archivo.get('filename')
            # Extraer la URL directa de descarga
            url_descarga = archivo.get('content_details', {}).get('download_url') 
            
            if not url_descarga or not nombre_archivo:
                continue
                
            ruta_local = os.path.join(DOWNLOAD_DIR, nombre_archivo)
            
            if os.path.exists(ruta_local):
                print(f"El archivo {nombre_archivo} ya existe localmente. Omitiendo...")
                continue
                
            tamano_mb = archivo.get('size', 0) / (1024 * 1024)
            print(f"Descargando {nombre_archivo} ({tamano_mb:.2f} MB)...")
            
            # Descarga en fragmentos para no saturar la memoria
            with requests.get(url_descarga, stream=True) as r:
                r.raise_for_status()
                with open(ruta_local, 'wb') as f:
                    for chunk in r.iter_content(chunk_size=8192):
                        f.write(chunk)
                        
            print(f"Guardado exitosamente: {ruta_local}")
            
    except Exception as e:
        print(f"Error durante la descarga: {e}")
        print(f"Alternativa: Descarga manual desde https://data.mendeley.com/datasets/{DATASET_ID}/{VERSION}")
```

```python
descargar_mendeley_v3()
print("\nProceso de descarga de HVDROPDB finalizado.")
```
